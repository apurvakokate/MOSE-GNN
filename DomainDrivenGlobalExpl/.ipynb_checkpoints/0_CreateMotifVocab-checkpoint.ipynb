```python
#─────────────────────────────────────────────────────────────────────
# UTILITY FUNCTIONS
#─────────────────────────────────────────────────────────────────────

def ensure_group_column(csv_path,
                        group_col: str = 'group',
                        train_frac: float = 0.8,
                        valid_frac: float = 0.1,
                        test_frac: float = 0.1,
                        random_state: int = 42) -> pd.DataFrame:
    """
    If `group_col` missing, does an 80/10/10 split,
    labels rows ('training','valid','test'), overwrites CSV, returns df.
    """
    df = pd.read_csv(csv_path)
    if group_col not in df.columns:
        # 1. split off test
        train_valid, test_df = train_test_split(df,
                                                test_size=test_frac,
                                                random_state=random_state)
        # 2. split train vs valid
        valid_size = valid_frac / (train_frac + valid_frac)
        train_df, valid_df = train_test_split(train_valid,
                                              test_size=valid_size,
                                              random_state=random_state)
        # 3. label & reassemble
        train_df[group_col] = 'training'
        valid_df[group_col] = 'valid'
        test_df[group_col]  = 'test'
        combined = pd.concat([train_df, valid_df, test_df], ignore_index=True)
        combined.to_csv(csv_path, index=False)
        return combined
    else:
        return df


def compute_length_stats(counts: Counter, lengths: dict) -> (float, float):
    """
    Given motif->normalized count and motif->length,
    return (mean_length, std_length).
    """
    total_weight = sum(counts.values())
    if total_weight == 0:
        return 0.0, 0.0

    # weighted mean
    mean_len = sum(counts[m] * lengths.get(m, 0)
                   for m in counts) / total_weight
    # weighted variance
    var = sum(counts[m] * (lengths.get(m, 0) - mean_len) ** 2
              for m in counts) / total_weight
    return mean_len, math.sqrt(var)

```

```python
# To Help User set the threshold
# Results storage
results = []
if True:
    dataset_size_dict = {}

    # Ensure output dir for plots
    os.makedirs("plots", exist_ok=True)

    # Threshold parameters
    IMBALANCE_MARGIN = 0.6               # for detecting minority class imbalance
    regression_datasets = ['Lipophilicity', 'esol']
    multi_label_datasets = ['tox21']

    # Main loop over datasets
    for dataset_name, label_col in CONSTANTS.DATASET_COLUMN.items():
        csv_path = f"./datasets/_{dataset_name}.csv"
        df = ensure_group_column(csv_path)

        if CONSTANTS.DATASET_TYPE[dataset_name] == 'BinaryClass':
            num_classes = 2
        elif CONSTANTS.DATASET_TYPE[dataset_name] == 'Regression':
            num_classes = 1
        else:
            num_classes = len(label_col)

        # Initialize splits
        training_data   = MolDataset(root=".", split='training', csv_file=csv_path, label_col=label_col)
        validation_data = MolDataset(root=".", split='valid',    csv_file=csv_path, label_col=label_col)
        test_data       = MolDataset(root=".", split='test',     csv_file=csv_path, label_col=label_col)

        # Combined for training+validation metrics
        combined_trainval = ConcatDataset([training_data, validation_data])
        N = len(combined_trainval)
        dataset_size_dict[dataset_name] = N

        # Flags
        is_regression = dataset_name in regression_datasets
        is_multilabel = dataset_name in multi_label_datasets

        # Graph-level labels for single-label datasets
        graph_labels = None
        if not is_multilabel and not is_regression:
            graph_labels = df.set_index('smiles').to_dict()
            n0 = sum(int(d.y.item() == 0) for d in combined_trainval)
            n1 = sum(int(d.y.item() == 1) for d in combined_trainval)

        # Loop over explanation algorithms
        for algorithm in CONSTANTS.ALGORITHMS:
            writers = {
                'global': imageio.get_writer(f'plots/{dataset_name}_{algorithm}_global.gif', mode='I', duration=15.0),
                'post':   imageio.get_writer(f'plots/{dataset_name}_{algorithm}_post.gif',   mode='I', duration=15.0),
                'length': imageio.get_writer(f'plots/{dataset_name}_{algorithm}_length.gif', mode='I', duration=15.0),
                'freq':   imageio.get_writer(f'plots/{dataset_name}_{algorithm}_freq.gif',   mode='I', duration=15.0)
            }

            # Loop over threshold percentages
            for pct in CONSTANTS.PERCENT_THRESHOLDS:

                # Build motif lookup tables
                data_lookup, test_lookup, lookup_obj = get_lookup_tables(
                    training_data, validation_data, test_data, algorithm=algorithm
                )

                # Motif counting per class. Skipped for regression
                if is_multilabel:
                    # Multi-label: count per task and label
                    num_tasks = num_classes
                    counts = {t: {0: Counter(), 1: Counter()} for t in range(num_tasks)}
                    for item in combined_trainval:
                        smiles = item.smiles
                        y = item.y.cpu().numpy().reshape(-1)
                        node2motif = data_lookup[smiles]
                        for motif in node2motif.values():
                            length = lookup_obj.motifs_length.get(motif, 1)
                            for t in range(num_tasks):
                                lbl = y[t]
                                if lbl in (0,1):
                                    counts[t][lbl][motif] += 1.0/length
                elif not is_regression:
                    # Binary or regression: count per class
                    counts = {0: Counter(), 1: Counter()}
                    for item in combined_trainval:
                        smiles = item.smiles
                        y = item.y.item()
                        node2motif = data_lookup[smiles]
                        for motif in node2motif.values():
                            length = lookup_obj.motifs_length.get(motif, 1)
                            counts[y][motif] += 1.0/length
    #                 for smiles, node2motif in data_lookup.items():

    #                     lbl = graph_labels[label_col[0]][smiles]
    #                     for motif in node2motif.values():
    #                         counts[lbl][motif] += 1.0/lookup_obj.motifs_length[motif]

                # Compute total weighted counts
                total_counts = Counter()
                for _, node2motif in data_lookup.items():
                    for motif in node2motif.values():
                        length = lookup_obj.motifs_length.get(motif, 1)
                        total_counts[motif] += 1.0 / length
                total_counts = dict(total_counts)

                global_cutoff = int((pct/100) * N)

                # All motifs in vocabulary before filtering
                all_motifs = list(lookup_obj.get_all_unique_motif())

                motif_global = [m for m, c in total_counts.items() if c >= global_cutoff]

                # Reindex global-only
                g2m_global = simplified_reindex_data(data_lookup, motif_global)
                g2m_global_test = simplified_reindex_data(test_lookup, motif_global)

                node_cov_global = compute_node_coverage(data_lookup)
                node_cov_global_test = compute_node_coverage(test_lookup)


                ######## 2) Post-minority selection (skip regression/multilabel) ###########
                motif_post = set(motif_global)
                if not is_regression and not is_multilabel:
                    r0, r1 = n0/N, n1/N
                    minority = 1 if r0>=IMBALANCE_MARGIN else (0 if r1>=IMBALANCE_MARGIN else None)
                    if minority is not None:
                        mb_cutoff = int((pct/100) * (n1 if minority==1 else n0))
                        for m, cnt in counts[minority].items():
                            if cnt >= mb_cutoff:
                                motif_post.add(m)
                motif_post = list(motif_post)

                # Prune vocabulary using post selection
                to_remove = [m for m in all_motifs if m not in motif_post]

                lookup_obj.remove_motifs(to_remove)

                motif_list = list(lookup_obj.get_all_unique_motif())
                assert(len(motif_list) == len(motif_post))
                # input(motif_list)

                # 2) Reindex train/val
                g2m_post = simplified_reindex_data(data_lookup, motif_list)


                # 3) Reindex test
                g2m_test   = simplified_reindex_data(test_lookup, motif_list)

                # Plot 1: global-only histogram
                fig = plt.figure();
                if is_multilabel or is_regression:
                    plt.hist([len(mots) for mots in g2m_global.values()], bins=range(0, max((len(m) for m in g2m_global.values()), default=0)+2), alpha=0.7)
                else:
                    grp0 = [len(g2m_global[g]) for g in g2m_global if graph_labels[label_col[0]][g]==0]
                    grp1 = [len(g2m_global[g]) for g in g2m_global if graph_labels[label_col[0]][g]==1]
                    plt.hist([grp0,grp1], bins=range(0, max(max(grp0 or [0]), max(grp1 or [0]))+2), stacked=True)
                plt.title(f"{pct}% global"); plt.tight_layout()
                buf = io.BytesIO(); fig.savefig(buf, format='png'); plt.close(fig); buf.seek(0)
                writers['global'].append_data(imageio.imread(buf)); buf.close()

                # Plot2: post-minority frame
                fig = plt.figure();
                if is_multilabel or is_regression:
                    plt.hist([len(mots) for mots in g2m_post.values()], bins=range(0, max((len(m) for m in g2m_post.values()), default=0)+2), alpha=0.7)
                else:
                    grp0 = [len(g2m_post[g]) for g in g2m_post if graph_labels[label_col[0]][g]==0]
                    grp1 = [len(g2m_post[g]) for g in g2m_post if graph_labels[label_col[0]][g]==1]
                    plt.hist([grp0,grp1], bins=range(0, max(max(grp0 or [0]), max(grp1 or [0]))+2), stacked=True)
                plt.title(f"{pct}% post-minority"); plt.tight_layout()
                buf = io.BytesIO(); fig.savefig(buf, format='png'); plt.close(fig); buf.seek(0)
                writers['post'].append_data(imageio.imread(buf)); buf.close()

                # Motif lengths frame
                lengths = [lookup_obj.motifs_length[m] for m in motif_post]
                fig = plt.figure(); plt.hist(lengths, bins=range(min(lengths or [0]), max(lengths or [0])+2)); plt.title(f'{pct}% lengths'); plt.tight_layout()
                buf = io.BytesIO(); fig.savefig(buf, format='png'); plt.close(fig); buf.seek(0)
                writers['length'].append_data(imageio.imread(buf)); buf.close()

                # Motif freqs frame
                freqs = [total_counts[m] for m in motif_post]
                fig = plt.figure(); plt.hist(freqs, bins=20); plt.title(f'{pct}% freqs'); plt.tight_layout()
                buf = io.BytesIO(); fig.savefig(buf, format='png'); plt.close(fig); buf.seek(0)
                writers['freq'].append_data(imageio.imread(buf)); buf.close()


                # Compute length stats
                mean_tv, std_tv = compute_length_stats(total_counts, lookup_obj.motifs_length)
                mean_test, std_test = compute_length_stats(total_counts, lookup_obj.motifs_length)

                # Record stats
                results.append({
                    'dataset': dataset_name,
                    'algorithm': algorithm,
                    'threshold_pct': pct,
                    'cutoff_count': global_cutoff,
                    'n_graphs_tv': N,
                    'n_graphs_test': len(test_lookup),
                    'vocab_size': len(motif_list),
                    'graphs_with_tv': len(g2m_post),
                    'graphs_with_test': len(g2m_test),
                    'graphs_with_tv_global': len(g2m_global),
                    'graphs_with_test_global': len(g2m_global_test),
                    'node_cov_tv': compute_node_coverage(data_lookup),
                    'node_cov_test': compute_node_coverage(test_lookup),
                    'vocab_size_global': len(motif_global),
                    'node_cov_tv_global': node_cov_global,
                    'node_cov_test_global': node_cov_global_test,
                    'mean_len_tv': mean_tv,
                    'std_len_tv': std_tv,
                    'mean_len_test': mean_test,
                    'std_len_test': std_test
                })
            for w in writers.values(): w.close()
```

```python
'''
Code for spliting data into 5 random fold.
These are set for fair evaluation across methods
'''
import pandas as pd
import numpy as np

# Make sure output directory exists
os.makedirs("datasets/FOLDS", exist_ok=True)

for random_state in [0, 1, 2, 3, 4]:
    for dataset_name in CONSTANTS.DATASET_COLUMN:
        fold_path = f"datasets/FOLDS/{dataset_name}_{random_state}.csv"

        # Skip if this fold file already exists
        if os.path.exists(fold_path):
            print(f"▶️  {fold_path} already exists—skipping.")
            continue

        # Load the original dataset CSV
        orig_path = f"datasets/_{dataset_name}.csv"
        df = pd.read_csv(orig_path)

        # Shuffle once with this seed
        df = df.sample(frac=1, random_state=random_state).reset_index(drop=True)

        # Compute split sizes
        n = len(df)
        train_end = int(0.8 * n)
        valid_end = train_end + int(0.1 * n)

        # Assign group labels
        df['group'] = (
            ['training'] * train_end +
            ['valid']    * (valid_end - train_end) +
            ['test']     * (n - valid_end)
        )

        # (Optional) reshuffle within groups
        df = df.sample(frac=1, random_state=random_state).reset_index(drop=True)

        # Write out the fold CSV
        df.to_csv(fold_path, index=False)
        print(df['group'].value_counts().to_dict(), f"→ saved to {fold_path}")
```

```python
# Results storage
results = []
dataset_size_dict = {}


# Main loop over datasets
if True:
    for dataset_name, label_col in CONSTANTS.DATASET_COLUMN.items():
        for fold in [0, 1, 2, 3, 4]:
            
            df = pd.read_csv(f"datasets/FOLDS/{dataset_name}_{fold}.csv")
            
            training_data = MolDataset(root=".", split='training',csv_file=f"datasets/FOLDS/{dataset_name}_{fold}.csv", label_col = label_col)
            validation_data = MolDataset(root=".", split='valid',csv_file=f"datasets/FOLDS/{dataset_name}_{fold}.csv", label_col = label_col)
            test_data = MolDataset(root=".", split='test',csv_file=f"datasets/FOLDS/{dataset_name}_{fold}.csv", label_col = label_col)

            if CONSTANTS.DATASET_TYPE[dataset_name] == 'BinaryClass':
                num_classes = 2
            elif CONSTANTS.DATASET_TYPE[dataset_name] == 'Regression':
                num_classes = 1
            else:
                num_classes = len(label_col)

            # Combined for training+validation metrics
            combined_trainval = ConcatDataset([training_data, validation_data])
            N = len(combined_trainval)
            dataset_size_dict[dataset_name] = N

            # Flags
            is_regression = dataset_name in regression_datasets
            is_multilabel = dataset_name in multi_label_datasets

            # Graph-level labels for single-label datasets
            graph_labels = None
            if not is_multilabel and not is_regression:
                graph_labels = df.set_index('smiles').to_dict()
                n0 = sum(int(d.y.item() == 0) for d in combined_trainval)
                n1 = sum(int(d.y.item() == 1) for d in combined_trainval)

            # Loop over explanation algorithms
            for algorithm in CONSTANTS.ALGORITHMS:
                
                pct = CHOSEN_THRESHOLD[algorithm][dataset_name]
                date_tag = None # Put 1225 or other customs tags here

                date_tag = f"{algorithm}{pct}" if date_tag is None else date_tag

                # Build motif lookup tables
                data_lookup, test_lookup, lookup_obj = get_lookup_tables(
                    training_data, validation_data, test_data, algorithm=algorithm
                )

                # Motif counting per class. Skipped for regression
                if is_multilabel:
                    # Multi-label: count per task and label
                    num_tasks = num_classes
                    counts = {t: {0: Counter(), 1: Counter()} for t in range(num_tasks)}
                    for item in combined_trainval:
                        smiles = item.smiles
                        y = item.y.cpu().numpy().reshape(-1)
                        node2motif = data_lookup[smiles]
                        for motif in node2motif.values():
                            length = lookup_obj.motifs_length.get(motif, 1)
                            for t in range(num_tasks):
                                lbl = y[t]
                                if lbl in (0,1):
                                    counts[t][lbl][motif] += 1.0/length
                elif not is_regression:
                    # Binary or regression: count per class
                    counts = {0: Counter(), 1: Counter()}
                    for item in combined_trainval:
                        smiles = item.smiles
                        y = item.y.item()
                        node2motif = data_lookup[smiles]
                        for motif in node2motif.values():
                            length = lookup_obj.motifs_length.get(motif, 1)
                            counts[y][motif] += 1.0/length

                # Compute total weighted counts
                total_counts = Counter()
                for _, node2motif in data_lookup.items():
                    for motif in node2motif.values():
                        length = lookup_obj.motifs_length.get(motif, 1)
                        total_counts[motif] += 1.0 / length
                total_counts = dict(total_counts)
            
                global_cutoff = int((pct/100) * N)

                # All motifs in vocabulary before filtering
                all_motifs = list(lookup_obj.get_all_unique_motif())

                motif_global = [m for m, c in total_counts.items() if c >= global_cutoff]
            

                ######## 2) Post-minority selection (skip regression/multilabel) ###########
                motif_post = set(motif_global)
                if not is_regression and not is_multilabel:
                    r0, r1 = n0/N, n1/N
                    minority = 1 if r0>=IMBALANCE_MARGIN else (0 if r1>=IMBALANCE_MARGIN else None)
                    if minority is not None:
                        mb_cutoff = int((pct/100) * (n1 if minority==1 else n0))
                        for m, cnt in counts[minority].items():
                            if cnt >= mb_cutoff:
                                motif_post.add(m)
                motif_post = list(motif_post)

                # Prune vocabulary using post selection
                to_remove = [m for m in all_motifs if m not in motif_post]

                lookup_obj.remove_motifs(to_remove)

                motif_list = list(lookup_obj.get_all_unique_motif())
                assert(len(motif_list) == len(motif_post))
                # input(motif_list)

                # 2) Reindex train/val
                g2m_post = simplified_reindex_data(data_lookup, motif_list)


                # 3) Reindex test
                g2m_test   = simplified_reindex_data(test_lookup, motif_list)
                
                test_mask_data = get_masked_graphs_from_list_for_each_motif(test_data, test_lookup, lookup_obj.get_motif_lengths())
                train_mask_data = get_masked_graphs_from_list_for_each_motif(training_data, data_lookup, lookup_obj.get_motif_lengths())
                val_mask_data = get_masked_graphs_from_list_for_each_motif(validation_data, data_lookup, lookup_obj.get_motif_lengths())

                write_to_file(data_lookup,dataset_name, fold, date_tag,'graph_lookup')
                write_to_file(list(lookup_obj.get_all_unique_motif()),dataset_name, fold, date_tag,'motif_list')
                write_to_file(lookup_obj.get_motif_lengths(),dataset_name, fold, date_tag,'motif_length')
                write_to_file(total_counts,dataset_name, fold, date_tag,'motif_counts')
                write_to_file(lookup_obj.motifs_class,dataset_name, fold, date_tag,'motif_class')
                write_to_file(g2m_post,dataset_name, fold, date_tag,'graph_motifidx')
                write_to_file(g2m_test,dataset_name, fold, date_tag,'test_graph_motifidx')
                write_to_file(test_lookup,dataset_name, fold, date_tag,'test_graph_lookup')
                write_to_file(lookup_obj.get_test_motif_lengths(),dataset_name, fold, date_tag,'test_graph_motif_length')
                write_to_file(test_mask_data,dataset_name, fold, date_tag,'test_dataset_masked')
                write_to_file(train_mask_data,dataset_name, fold, date_tag,'train_dataset_masked')
                write_to_file(val_mask_data,dataset_name, fold, date_tag,'validation_dataset_masked')

            
```

```python
import csv
import os

csv_output_dir = f"plots/csv_exports"
os.makedirs(csv_output_dir, exist_ok=True)

# Main loop over datasets
if True:
    for dataset_name, label_col in CONSTANTS.DATASET_COLUMN.items():
        for fold in [0, 1, 2, 3, 4]:
            
            df = pd.read_csv(f"datasets/FOLDS/{dataset_name}_{fold}.csv")
            
            training_data = MolDataset(root=".", split='training',csv_file=f"datasets/FOLDS/{dataset_name}_{fold}.csv", label_col = label_col)
            validation_data = MolDataset(root=".", split='valid',csv_file=f"datasets/FOLDS/{dataset_name}_{fold}.csv", label_col = label_col)
            test_data = MolDataset(root=".", split='test',csv_file=f"datasets/FOLDS/{dataset_name}_{fold}.csv", label_col = label_col)

            if CONSTANTS.DATASET_TYPE[dataset_name] == 'BinaryClass':
                num_classes = 2
            elif CONSTANTS.DATASET_TYPE[dataset_name] == 'Regression':
                num_classes = 1
            else:
                num_classes = len(label_col)

            # Combined for training+validation metrics
            combined_trainval = ConcatDataset([training_data, validation_data])
            N = len(combined_trainval)
            dataset_size_dict[dataset_name] = N

            # Flags
            is_regression = dataset_name in regression_datasets
            is_multilabel = dataset_name in multi_label_datasets

            # Graph-level labels for single-label datasets
            graph_labels = None
            if not is_multilabel and not is_regression:
                graph_labels = df.set_index('smiles').to_dict()
                n0 = sum(int(d.y.item() == 0) for d in combined_trainval)
                n1 = sum(int(d.y.item() == 1) for d in combined_trainval)

            # Loop over explanation algorithms
            for algorithm in CONSTANTS.ALGORITHMS:
                
                csv_rows = []
                
                pct = CHOSEN_THRESHOLD[algorithm][dataset_name]
                date_tag = None # Put 1225 or other customs tags here

                date_tag = f"{algorithm}{pct}" if date_tag is None else date_tag

                # Build motif lookup tables
                data_lookup, test_lookup, lookup_obj = get_lookup_tables(
                    training_data, validation_data, test_data, algorithm=algorithm
                )
                
                for smiles, node2motif in data_lookup.items():
                    label_entry = df[df['smiles'] == smiles]
                    if label_entry.empty:
                        continue

                    # Handle label based on task type
                    if is_regression:
                        label = label_entry[label_col[0]].values[0]
                    elif is_multilabel:
                        label = label_entry[label_col].iloc[0].to_numpy()
                    else:
                        label = int(label_entry[label_col[0]].values[0])

                    motif_counts_local = Counter()
                    for motif in node2motif.values():
                        length = lookup_obj.motifs_length.get(motif, 1)
                        motif_counts_local[motif] += 1.0 / length

                    for motif_str, freq in motif_counts_local.items():
                        row = {
                            "Motif String": motif_str,
                            "Motif Length": lookup_obj.motifs_length.get(motif_str, 1),
                            "Motif Frequency": freq,
                            "Graph String": smiles,
                            "Label": label
                        }
                        csv_rows.append(row)

                # Write full CSV for this dataset/fold
                csv_output_file = os.path.join(csv_output_dir, f"{dataset_name}_{fold}_{algorithm}.csv")
                with open(csv_output_file, "w", newline='') as f:
                    writer = csv.DictWriter(f, fieldnames=["Motif String", "Motif Length", "Motif Frequency", "Graph String", "Label"])
                    writer.writeheader()
                    writer.writerows(csv_rows)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import plotly.express as px

def interactive_motif_matrix(csv_file):
    df = pd.read_csv(csv_file)
    df['Motif Length'] = df['Motif Length'].astype(int)

    df_class0 = df[df['Label'] == 0]
    df_class1 = df[df['Label'] == 1]

    motifs_sorted = df[['Motif String', 'Motif Length']].drop_duplicates().sort_values(by='Motif Length')
    motifs = motifs_sorted['Motif String'].tolist()

    graphs_class0 = df_class0['Graph String'].unique().tolist()
    graphs_class1 = df_class1['Graph String'].unique().tolist()

    matrix0 = pd.DataFrame(0, index=motifs, columns=graphs_class0)
    matrix1 = pd.DataFrame(0, index=motifs, columns=graphs_class1)

    for _, row in df_class0.iterrows():
        matrix0.at[row['Motif String'], row['Graph String']] = 1

    for _, row in df_class1.iterrows():
        matrix1.at[row['Motif String'], row['Graph String']] = 1

    # Combine and mark class
    df0 = matrix0.reset_index().melt(id_vars='index', var_name='Graph', value_name='Presence')
    df0['Class'] = 'Class 0'

    df1 = matrix1.reset_index().melt(id_vars='index', var_name='Graph', value_name='Presence')
    df1['Class'] = 'Class 1'

    full_df = pd.concat([df0, df1], axis=0)
    full_df = full_df[full_df['Presence'] == 1]

    fig = px.scatter(
        full_df, x='Graph', y='index', color='Class',
        title='Motif Presence in Graphs by Class',
        labels={'index': 'Motif', 'Graph': 'Graph ID'},
        height=1000
    )
    fig.update_layout(yaxis={'categoryorder':'array', 'categoryarray':motifs})
    fig.show()
```

```python
# Example usage
csv_path = "csv_exports/Mutagenicity_0_RBRICS.csv"  # Change as needed
interactive_motif_matrix(csv_path)
```

```python
def plot_motif_bias(csv_file, min_support=5):
    # Load and prepare
    df = pd.read_csv(csv_file)
    df['Motif Length'] = df['Motif Length'].astype(int)
    
    # Deduplicate motifs within each graph
    df_unique = df.drop_duplicates(subset=['Graph String', 'Motif String'])


    # Group motif occurrences by class
    counts = df_unique.groupby(['Motif String', 'Label']).size().unstack(fill_value=0)

    # Add motif length from original df
    motif_lengths = df.groupby('Motif String')['Motif Length'].first()
    counts['Motif Length'] = motif_lengths

    # Calculate total and class 1 ratio
    counts['total'] = counts.sum(axis=1)
    counts['class1_ratio'] = counts.get(1, 0) / counts['total']

    # Filter out motifs with too few total appearances
    counts = counts[counts['total'] >= min_support]

    # Plot with color by motif length
    plt.figure(figsize=(10, 6))
    sc = plt.scatter(
        counts['total'],
        counts['class1_ratio'],
        c=counts['Motif Length'],
        cmap='viridis',
        alpha=0.8,
        edgecolor='k'
    )
    plt.colorbar(sc, label='Motif Length')
    plt.axhline(0.5, color='gray', linestyle='--', label='Balanced (50%)')
    plt.xlabel('Motif Frequency (Total Graphs)')
    plt.ylabel('Fraction in Class 1')
    plt.title('Motif Frequency vs. Class Bias (Colored by Motif Length)')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt


def plot_motif_bias_and_list(csv_file, min_support=5):
    # Load data
    df = pd.read_csv(csv_file)
    df['Motif Length'] = df['Motif Length'].astype(int)

    # Group by motif + label
    counts = df.groupby(['Motif String', 'Label']).size().unstack(fill_value=0)
    motif_lengths = df.groupby('Motif String')['Motif Length'].first()
    counts['Motif Length'] = motif_lengths

    # Metrics
    counts['total'] = counts.sum(axis=1)
    counts['class1_ratio'] = counts.get(1, 0) / counts['total']
    counts = counts[counts['total'] >= min_support]

    # Plot
    plt.figure(figsize=(10, 6))
    sc = plt.scatter(
        counts['total'],
        counts['class1_ratio'],
        c=counts['Motif Length'],
        cmap='viridis',
        alpha=0.8,
        edgecolor='k'
    )
    plt.colorbar(sc, label='Motif Length')
    plt.axhline(0.5, color='gray', linestyle='--', label='Balanced (50%)')
    plt.xlabel('Motif Frequency (Total Graphs)')
    plt.ylabel('Fraction in Class 1')
    plt.title('Motif Frequency vs. Class Bias (Colored by Motif Length)')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.legend()
    plt.tight_layout()
    plt.show()

    # Print most biased motifs
    top_class0 = counts.sort_values('class1_ratio').head(5)
    top_class1 = counts.sort_values('class1_ratio', ascending=False).head(5)

    print("\nTop 5 Motifs Biased Toward Class 0:")
    print(top_class0[['total', 'class1_ratio', 'Motif Length']])

    print("\nTop 5 Motifs Biased Toward Class 1:")
    print(top_class1[['total', 'class1_ratio', 'Motif Length']])
```

```python
import pandas as pd
import os
import glob

# Directory containing CSV files
csv_folder = f"{folder_name}/plots/motif_csvs"  # Change this path if needed

# File pattern
csv_files = glob.glob(os.path.join(csv_folder, "*_*_top10.csv")) + \
            glob.glob(os.path.join(csv_folder, "*_*_bottom10.csv"))

# Columns to retain
columns_to_keep = [
    "motif", "avg_sigmoid_importance", "median_logit_diff", "mode_sign",
    "mode_logit_diff", "avg_logit_diff", "max_logit_diff", "duplicate_counts",
    "frequency", "max_logit_diff_graph"
]

all_data = []

for file in csv_files:
    try:
        filename = os.path.basename(file).replace(".csv", "")
        parts = filename.split("_")

        # Parse dataset name, architecture, and rank type
        if parts[-1] in {"top10", "bottom10"}:
            label = parts[-1]
            architecture = parts[-2]
            dataset = "_".join(parts[:-2])
        else:
            print(f"Skipping invalid filename format: {filename}")
            continue

        df = pd.read_csv(file)

        # Ensure required columns are present
        if not set(columns_to_keep).issubset(df.columns):
            print(f"Skipping {filename}: missing required columns")
            continue

        # Select and round float columns
        df = df[columns_to_keep]
        float_cols = df.select_dtypes(include=['float']).columns
        df[float_cols] = df[float_cols].round(3)

        # Add metadata
        df["dataset"] = dataset
        df["architecture"] = architecture
        df["rank_type"] = label

        all_data.append(df)
    except Exception as e:
        print(f"Error processing {file}: {e}")

# Combine all data
if all_data:
    combined_df = pd.concat(all_data, ignore_index=True)
    combined_df.to_csv("combined_motif_stats.csv", index=False)
    print("Saved combined_motif_stats.csv")
else:
    print("No valid data to combine.")
```

```python
import os
import numpy as np
import pandas as pd
import torch

# ensure output directory
os.makedirs("plots", exist_ok=True)

# storage for every metric
summary_rows = []

def analyze_split(mol_dataset, split_name, dataset_name, fold):
    n = len(mol_dataset)
    # extract all labels into an array
    ys = []
    for i in range(n):
        y = mol_dataset[i].y
        ys.append(y.cpu().numpy().reshape(-1))
    ys = np.vstack(ys)  # (n, k)

    # 1) TARGET DISTRIBUTION / STATS
    ds_type = CONSTANTS.DATASET_TYPE[dataset_name]
    if ds_type == 'BinaryClass' or ds_type.startswith('MultiClass'):
        # single‐label classification
        vals, counts = np.unique(ys[:,0], return_counts=True)
        for val, cnt in zip(vals, counts):
            summary_rows.append({
                'dataset': dataset_name,
                'fold':    fold,
                'split':   split_name,
                'metric':  f'label_{int(val)}_count',
                'value':   int(cnt)
            })
            summary_rows.append({
                'dataset': dataset_name,
                'fold':    fold,
                'split':   split_name,
                'metric':  f'label_{int(val)}_pct',
                'value':   100 * cnt / n
            })

    elif ds_type == 'Regression':
        arr = ys[:,0]
        stats = {
            'reg_mean':    arr.mean(),
            'reg_std':     arr.std(),
            'reg_min':     arr.min(),
            'reg_25%':     np.percentile(arr,25),
            'reg_50%':     np.percentile(arr,50),
            'reg_75%':     np.percentile(arr,75),
            'reg_max':     arr.max()
        }
        for m,v in stats.items():
            summary_rows.append({
                'dataset': dataset_name,
                'fold':    fold,
                'split':   split_name,
                'metric':  m,
                'value':   v
            })

    else:
        # multi‐label classification
        label_sums = ys.sum(axis=0).astype(int)
        for idx, cnt in enumerate(label_sums):
            summary_rows.append({
                'dataset': dataset_name,
                'fold':    fold,
                'split':   split_name,
                'metric':  f'label_{idx}_pos_count',
                'value':   int(cnt)
            })
            summary_rows.append({
                'dataset': dataset_name,
                'fold':    fold,
                'split':   split_name,
                'metric':  f'label_{idx}_pos_pct',
                'value':   100 * cnt / n
            })

    # 2) GRAPH SIZE DISTRIBUTION
    node_counts = []
    edge_counts = []
    for i in range(n):
        g = mol_dataset[i]
        node_counts.append(g.num_nodes)
        edge_counts.append(
            (g.edge_index.shape[1] // 2) if hasattr(g, 'edge_index')
            else getattr(g, 'num_edges', 0)
        )

    def summarize(arr, name):
        return {
            f'{name}_min':    np.min(arr),
            f'{name}_25%':    np.percentile(arr,25),
            f'{name}_median': np.median(arr),
            f'{name}_75%':    np.percentile(arr,75),
            f'{name}_max':    np.max(arr),
            f'{name}_mean':   np.mean(arr),
            f'{name}_std':    np.std(arr),
        }

    for m,v in summarize(node_counts, 'nodes').items():
        summary_rows.append({
            'dataset': dataset_name,
            'fold':    fold,
            'split':   split_name,
            'metric':  m,
            'value':   v
        })
    for m,v in summarize(edge_counts, 'edges').items():
        summary_rows.append({
            'dataset': dataset_name,
            'fold':    fold,
            'split':   split_name,
            'metric':  m,
            'value':   v
        })


# run analysis
folds = [0,1,2,3,4]
for dataset_name, label_col in CONSTANTS.DATASET_COLUMN.items():
    for fold in folds:
        for split in ['training','valid','test']:
            ds = MolDataset(
                root=".", split=split,
                csv_file=f"datasets/FOLDS/{dataset_name}_{fold}.csv",
                label_col=label_col, normalize=False
            )
            analyze_split(ds, split, dataset_name, fold)

# write out a single long‐form table
df_summary = pd.DataFrame(summary_rows)
out_path   = "plots/data_summary.csv"
df_summary.to_csv(out_path, index=False)
print(f"✅ Saved summary table to {out_path}")
```
